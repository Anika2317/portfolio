        <li style="margin-bottom: 4px; color: #ffffff;"><b style="color: #a3d9b5; font-weight: 900;">How it works:</b> String inputs from interface text fields are explicitly transformed into real numerical data using the global `Number()` method.
```python
// EVIDENCE: DATA TYPES - NUMBERS
// Source: Variables Homework & Math Expressions Homework

// Part 1: Type Conversion and Basic Arithmetic
function calculateMagic() {
    let input = document.getElementById("magicInput"); // Retrieves input element
    
    // Explicit Type Conversion: Transforming a string input value into a Number data type
    let magicNumber = Number(input.value);

    // Using Arithmetic Operators (* for multiplication, ** for exponentiation, / for division)
    let doubled = magicNumber * 2;
    let squared = magicNumber ** 2; // Exponentiation operator
    let half = magicNumber / 2;     // Division operator

    return { doubled, squared, half };
}

// Part 2: Isolated Math Expression Operations
function MathOperationsReview() {
    let num1 = 6; // Numeric property assignment (Integer)
    let num2 = 3; // Numeric property assignment (Integer)

    // Complex operator precedence evaluation
    let part1_result = (num1 * num2) + (num1 / 2); // Evaluates to 21
    
    // Storing calculated equation values back into a numeric variable
    let total = (num1 / num2) + 2 ** 3;            // Evaluates to 10

    return { part1_result, total };
}
```

```python
// EVIDENCE: DATA TYPES - STRINGS
// Source: Variables Homework

function StringManipulationReview() {
    // String Literals: Initializing string sequences using double quotes
    let firstName = "Anika";
    let lastName = "Seksaria";
    let favoriteFood = "pizza";

    // String Concatenation: Joining string variables and text literals using the + operator
    let fullName = firstName + " " + lastName;
    let baselineSentence = "My favorite food is " + favoriteFood + ".";

    // Template Literals: Using backticks to embed dynamic variables inside a string template
    let userName = "Anika";
    let userAge = "14";
    let templateOutput = `Hello ${userName}, you are ${userAge} years old.`;

    return { fullName, baselineSentence, templateOutput };
}
```

```python
// EVIDENCE: DATA TYPES - BOOLEANS
// Source: Red Riding Hood Game Architecture

class GameLevelRedRidingHood4 {
    constructor(gameEnv) {
        this.gameEnv = gameEnv;

        // Boolean Flags: Pure true/false variables used to track game states
        this.gameOver = false;       // Toggled to true when player health hits zero or timer expires
        this.isPaused = false;       // Boolean switch tracking user input pause requests
        this.enemyDefeated = false;  // Flag monitoring boss removal requirements
    }

    update() {
        // Boolean Logic Check: Bypasses execution blocks entirely if a state evaluates to true
        if (this.gameOver || this.isPaused) return;

        const player = this.gameEnv.gameObjects.find(obj => obj instanceof ShooterPlayer);
        if (!player) return;

        // Utilizing Boolean Flags: Evaluating custom state fields on dynamic entities
        if (player.isJumping) {
            player.applyGravityPhysics();
        }

        if (player.isVulnerable) {
            this.checkEnemyCollisions(player);
        }
    }
}
```

        <li style="color: #ffffff;"><b style="color: #a3d9b5; font-weight: 900;">Why it helps:</b> It prevents variable bloat.
```python
// EVIDENCE: DATA TYPES - ARRAYS
// Source: Red Riding Hood Game Architecture

class GameLevelRedRidingHood4 {
    constructor(gameEnv) {
        this.gameEnv = gameEnv;

        // Array Declaration: An ordered collection holding dynamic projectile elements
        this.activeBullets = [];

        // Array Declaration: Grouping multiple image background file paths under one reference
        this.backgroundLayers = [
            "/images/projects/red-riding/layer1-sky.png",
            "/images/projects/red-riding/layer2-trees.png",
            "/images/projects/red-riding/layer3-ground.png"
        ];
    }

    spawnBullet(x, y) {
        const newBullet = new Bullet(x, y);
        
        // Array Operation: Appending a newly created object to the end of the collection
        this.activeBullets.push(newBullet);
    }

    update() {
        // Array Operations: Filtering out and removing destroyed bullets from the active tracking loop
        this.activeBullets = this.activeBullets.filter(bullet => !bullet.destroyed);
    }
}
```

```python
// EVIDENCE: DATA TYPES - JSON OBJECTS
// Source: Red Riding Hood Game Architecture

class GameLevelRedRidingHood4 {
    constructor(gameEnv) {
        this.gameEnv = gameEnv;

        // JSON Object: Structured configuration schema mapping game asset metadata
        this.wolfAssetConfig = {
            "id": "boss_wolf_01",
            "name": "Big Bad Wolf",
            "isHostile": true,
            "dimensions": {
                "width": 632,
                "height": 395,
                "scaleMultiplier": 3.0
            },
            "animations": {
                "idle": "/images/projects/red-riding/wolf-idle.png",
                "attack": "/images/projects/red-riding/wolf-attack.png"
            }
        };
    }

    initializeEnemy() {
        // Parsing property details from the nested JSON structural setup using dot notation
        const spritePath = this.wolfAssetConfig.animations.attack;
        const collisionWidth = this.wolfAssetConfig.dimensions.width * this.wolfAssetConfig.dimensions.scaleMultiplier;
        
        console.log(`Loading asset configuration for: ${this.wolfAssetConfig.name}`);
        
        // Factory creation pattern using structural data configuration fields
        return new Enemy(spritePath, collisionWidth);
    }
}
```
