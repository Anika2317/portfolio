## Popcorn Hack 2

```python
# CODE_RUNNER: Lists Popcorn 2

numbers = [7, 6, 8, 5, 7, 9] # Create the list of numbers provided in the lesson

total = 0 # Initialize a starting variable at 0 to keep track of our running sum

for num in numbers: # Start a loop that looks at every individual number in the list one by one
    total = total + num # Take the current number and add it to our running total

print(total) # Display the final calculated sum, which should be 42

average = total / len(numbers) # Calculate the average by dividing the total sum by how many numbers there are
print(average) # Display the calculated average, which should be 7.0
```

## Popcorn Hack 3

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]   # Store the class voting data

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1


# 1. Build a list of items with 2 or more votes
popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

# 2. Print that list
print(popular)
```

## Homework

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset
votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips", "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]
prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
# 1. Print how many votes were cast
print(len(votes))

# 2. Print the last vote using indexing
print(votes[-1])

# LESSON 2 - sum and average
# 3. Use a loop to total prices (do NOT use sum())
total = 0
for price in prices:
    total = total + price

# 4. Print "Total: <total>" and "Average: <average>" as integers using the int() function
average = total / len(prices)
print("Total:", int(total))
print("Average:", int(average))

# LESSON 3 - count and filter
# 5. Build a frequency dictionary of votes and print it
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1
print(frequency)

# 6. Build a list of items with 3 or more votes and print it
top_snacks = []
for item in frequency:
    if frequency[item] >= 3:
        top_snacks.append(item)
print(top_snacks)
```
