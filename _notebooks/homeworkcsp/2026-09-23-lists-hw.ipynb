```python
# CODE_RUNNER: Lists Popcorn 2 

numbers = [7, 6, 8, 5, 7, 9] # Create the list of numbers provided in the lesson

total = 0 # Initialize a starting variable at 0 to keep track of our running sum

for num in numbers: # Start a loop that looks at every individual number in the list one by one
    total = total + num # Take the current number and add it to our running total

print(total) # Display the final calculated sum, which should be 42

if len(numbers) == 0: # Check if the list is empty so we don't accidentally divide by zero
    print("No data yet") # Display a warning if the list is empty
else: # If the list actually has numbers inside of it
    average = total / len(numbers) # Calculate the average by dividing the total sum by how many numbers there are
    print(average) # Display the calculated average, which should be 7.0
```

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"] # Store the class voting data

frequency = {} # Create an empty dictionary where we will store each season and its vote count
for season in seasons: # Start a loop to check every single vote in our list
    if season in frequency: # Check if we have already started counting this specific season
        frequency[season] += 1 # If we have, add 1 to its current vote count
    else: # If this is the very first time we are seeing this season
        frequency[season] = 1 # Add it to our dictionary and set its starting count to 1

popular = [] # Create a brand new, empty list to hold only the seasons that got a lot of votes
for season in frequency: # Loop through every unique season currently stored in our dictionary
    if frequency[season] >= 2: # Check if the vote count for this season is greater than or equal to 2
        popular.append(season) # If the condition is true, add this season's name into our popular list

print(popular) # Display the final filtered list, which will just be ['summer', 'fall']
```

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips", "pretzels", "cookies", "chips", "fruit", "chips", "cookies"] # Create a list of all the snack votes
prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3] # Create a matching list of the prices for each vote

# LESSON 1 - indexing
print(len(votes)) # Print the total number of votes cast by getting the length of the list (12)
print(votes[-1]) # Print the very last vote in the list using the -1 index ("cookies")

# LESSON 2 - sum and average
total = 0 # Create a variable starting at 0 to keep track of the total price
for price in prices: # Loop through every single number inside the prices list
    total = total + price # Add the current price to our running total

average = total / len(prices) # Calculate the average by dividing the total money spent by the number of items bought

print("Total:", int(total)) # Print the text "Total:" followed by the total sum converted to a whole integer (31)
print("Average:", int(average)) # Print the text "Average:" followed by the average converted to a whole integer (2)

# LESSON 3 - count and filter
frequency = {} # Create an empty dictionary to act as our vote counter
for vote in votes: # Loop through every single vote in the votes list
    if vote in frequency: # Check if we already have this snack in our counter
        frequency[vote] += 1 # If yes, increase its count by 1
    else: # If we haven't seen this snack yet
        frequency[vote] = 1 # Add it to the dictionary and set its vote count to 1

print(frequency) # Print out the entire dictionary showing the final counts for each snack

top_snacks = [] # Create an empty list to store snacks that received 3 or more votes
for item in frequency: # Loop through every unique snack inside our dictionary counter
    if frequency[item] >= 3: # Check if the vote count for this specific snack is 3 or higher
        top_snacks.append(item) # If it got 3 or more votes, add the snack's name to our top_snacks list

print(top_snacks) # Print the final filtered list of top snacks (['chips', 'cookies'])
```
