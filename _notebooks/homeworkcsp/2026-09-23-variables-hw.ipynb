```python
## Popcorn
stars = 8 # Set the initial value of the stars variable to 8
checkpoint_stars = stars # Store the current value of stars (8) into a new variable called checkpoint_stars
stars = stars + 4 # Add 4 to the current value of stars (8 + 4) and save it back to the stars variable (now 12)

print(stars) # Display the current value of the stars variable, which is 12
print(checkpoint_stars) # Display the value stored in checkpoint_stars, which remains 8
```

```python
# UESL Prototype
game_title = "Little Red Riding Hood" # Create a variable to store the text name of the game
reduced_motion = True # Create a Boolean variable to store the player's preference for reduced motion (True means it is on)

level_one_stars = 12 # Create a variable for level 1 and set its star count to 12
level_two_stars = 18 # Create a variable for level 2 and set its star count to 18
level_three_stars = 10 # Create a variable for level 3 and set its star count to 10

# Calculate initial total
total_stars = level_one_stars + level_two_stars + level_three_stars # Add the stars from all three levels together (12 + 18 + 10) and store the result (40) in total_stars

# Print player profile
print(game_title) # Display the name of the game ("Little Red Riding Hood")
print(reduced_motion) # Display the player's reduced motion setting (True)
print(total_stars) # Display the initial calculated total of stars (40)

# Award bonus to level two
level_two_stars = level_two_stars + 5 # Add a 5-star bonus to the current level 2 stars (18 + 5) and update the level_two_stars variable (now 23)

# Print stored total (before recalculation)
print(total_stars) # Display the total_stars variable again, which is still storing the old value of 40 because we haven't recalculated it yet

# Recalculate and print updated total
total_stars = level_one_stars + level_two_stars + level_three_stars # Add the levels together again using the new level 2 value (12 + 23 + 10) to update total_stars
print(total_stars) # Display the newly calculated total of stars (45)
```
