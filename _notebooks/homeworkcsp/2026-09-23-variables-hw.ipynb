### Design Thinking

Environment: To manage visual accessibility, I implemented a `reduced_motion` Boolean variable that simply flips between True and False.

Revision: I explicitly instructed the program to run the addition formula a second time (`total_apples = forest_apples + river_apples + village_apples`) immediately after the river stage bonus was given so the UI data refreshes.
