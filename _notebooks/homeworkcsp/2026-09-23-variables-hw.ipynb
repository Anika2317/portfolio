## Design Thinking

**Desires:** Players navigating the forest to Grandma's house need clear, accurate feedback on the items they have gathered. They also require an accessible user interface that is easy to read regardless of their visual capabilities.

**Enviornment:** To manage visual accessibility, I implemented a `colorblind_mode` Boolean variable that simply flips between `True` and `False`. I originally thought about using a text input where users type their specific color deficiency (like "Deuteranomaly" or "Protanopia"), but a straightforward on/off toggle that adjusts the game to a universal high-contrast palette is much cleaner to program and instantly usable.

**Prototype:** This foundational script models the core logic. It sets up the game's title, visual settings, and individual stage item counts, simulates an item increase during gameplay, and demonstrates the coding trap of "stale variables" by showing the old total next to the properly recalculated total.

**Colorblind mode as a player choice:** Accessibility features like colorblind mode are crucial because games heavily rely on color cues—such as a red warning flash when the wolf is near or green glowing paths for safety. If players cannot distinguish these colors, they miss vital gameplay information. Allowing users to toggle this setting ensures the game swaps critical color cues for distinct patterns or high-contrast icons, making it inclusive and playable for everyone without changing the actual gameplay difficulty.
