## Homework

**Explanation of the Abstraction:** 
By using `level_stars = [12, 18, 10]`, each element in the list represents one specific level’s collected stars, organized chronologically in level order.

```python
# Refactored 3.1 prototype into a list
level_stars = [12, 18, 10]

# Add 5 to the second level (Python index 1)
level_stars[1] = level_stars[1] + 5

# Append 7 for a new level
level_stars.append(7)

# Traversal to calculate total stars
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
```

## Design Thinking

* **Player/Maker Need:** The game maker wants to add, reorder, or remove levels efficiently without losing track of stars or rewriting the calculation logic.

* **Goal:** Create one understandable collection in chronological level order that automatically updates lengths and totals.

* **Representation Considered:** Used an ordered list of integers (`[12, 18, 10]`) so mathematical operations (like calculating the total) function properly across all levels.

* **Prototype:** Created a Python script with a single list and a `for` loop to track the length and calculate the total score dynamically.

* **Test/Revision:** I tested the traversal loop with an empty list `[]` to ensure new games start cleanly at 0 levels and 0 stars instead of crashing. 

* **Future Suggestion:** Incorporate clear level labels (e.g., "Forest", "Cavern") using dictionaries or parallel lists so players see names instead of relying on index numbers they may not understand.
