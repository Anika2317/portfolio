## Homework
Language: Python

```python
# CODE_RUNNER: Library 3.17 Homework - Compare catalog growth
def check_growth(n):
    visits = n
    pairs = n * n
    print(f"Books: {n}")
    print(f"Visits (Linear): {visits}")
    print(f"Pairs (Quadratic): {pairs}\n")

check_growth(12)
check_growth(24)
```

Prediction:
For n = 12: expect 12, 12, 144.
For n = 24: expect 24, 24, 576.

Actual Output:
- `Books: 12`
- `Visits (Linear): 12`
- `Pairs (Quadratic): 144`

Books: 24
Visits (Linear): 24
Pairs (Quadratic): 576

Explanation of Growth:
Visits grow linearly (doubling 12 to 24 doubles visits). Pairs grow quadratically (doubling n quadruples the pairs from 144 to 576).

Recommendation (1,000 Books):
For 1,000 books, linear search takes 1,000 checks (worst case); binary search takes only 10. If the catalog is sorted, use binary search. If it is unsorted, you must use linear search or sort it first (which requires extra work or a separate stored copy).
