## Homework Hack: SFI Backend Validator

```python
# CODE_RUNNER: Homework Hack - SFI Backend Validator
existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    }
]

# Build and test your validator below.
def validate_record(record, existing_specs, categories):
    has_name = record["product_name"] != ""
    valid_cat = record["category"] in categories
    has_spec = record["spec_number"] != ""
    has_date = record["effective_date"] != ""
    duplicate_spec = record["spec_number"] in existing_specs

    # Every requirement must pass (and), and the spec must NOT be a duplicate (not)
    return has_name and valid_cat and has_spec and has_date and not duplicate_spec

for record in test_records:
    if validate_record(record, existing_spec_numbers, valid_categories):
        print("ACCEPT RECORD:", record["product_name"])
    else:
        print("REJECT RECORD:", record["product_name"])
```

Homework output:
ACCEPT RECORD: Multiple Disc Clutch Assemblies
REJECT RECORD: Replacement Flywheels
REJECT RECORD: Existing Flywheel Record

Homework tested: The reusable validator correctly iterated through the 3 cases. 
1. Accepted because it meets all criteria and spec 1.2 is new. 
2. Rejected because "Street Car" is not in valid_categories.
3. Rejected because spec "1.1" is already in existing_spec_numbers (duplicate).

Boolean logic: `validate_record()` builds one small Boolean per rule, then combines them with `and`, because **all** requirements must pass. A single False (like `valid_cat` for "Street Car") makes the whole result False. The duplicate rule uses `not duplicate_spec`, so a spec already in `existing_spec_numbers` flips to False and rejects the record. Because it's a procedure with parameters, the same validator can be reused on any record or spec list.
