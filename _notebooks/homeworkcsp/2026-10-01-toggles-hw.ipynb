```python
%%html
<!-- UI_RUNNER: Toggles Homework Base -->

<div id="hw-container" style="padding: 20px; transition: 0.3s; background: white; color: black; border-radius: 8px; border: 1px solid #ccc;">

  <h3>User Profile Settings</h3>
  
  <!-- Toggle 1: Dark Mode -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle-dark" class="ocs__toggle-input hw-toggle" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Enable Dark Mode</span>
  </label>
  <br><br>

  <!-- Toggle 2: Show/Hide Avatar -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle-avatar" class="ocs__toggle-input hw-toggle" type="checkbox" checked>
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show Profile Avatar</span>
  </label>
  <br><br>

  <!-- Toggle 3: Show/Hide Bio -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="toggle-bio" class="ocs__toggle-input hw-toggle" type="checkbox" checked>
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show User Bio</span>
  </label>
  
  <hr>

  <!-- Active Toggle Counter -->
  <p><strong>Active Toggles: <span id="active-count">2</span>/3</strong></p>

  <hr>
  
  <!-- Live Features Controlled by the Toggles -->
  <h4>Profile Preview</h4>
  <div id="avatar-container" style="width: 60px; height: 60px; background: #3B82F6; border-radius: 50%; display: block;"></div>
  <p id="bio-container" style="display: block;">Hello! I am a student learning AP CSP and how to use SASS toggle grammar to manipulate the DOM.</p>

</div>

<script>
  // COMMENT 1: I learned that using document.querySelectorAll creates a NodeList, which lets us grab all elements with a specific class and loop through them easily!
  const toggles = document.querySelectorAll(".hw-toggle");
  const countDisplay = document.getElementById("active-count");
  
  const container = document.getElementById("hw-container");
  const toggleDark = document.getElementById("toggle-dark");
  const toggleAvatar = document.getElementById("toggle-avatar");
  const toggleBio = document.getElementById("toggle-bio");
  
  const avatarContainer = document.getElementById("avatar-container");
  const bioContainer = document.getElementById("bio-container");

  // COMMENT 2: The 'change' event listener is super useful because it automatically fires whenever the toggle is clicked and the state updates, preventing the need for manual button clicks.
  toggles.forEach(toggle => {
    toggle.addEventListener("change", () => {
      
      // REQUIREMENT: Calculate how many toggles are checked
      let activeCount = 0;
      toggles.forEach(t => {
        if (t.checked) activeCount++;
      });
      // Update the HTML text with the current count
      countDisplay.textContent = activeCount;

      // REQUIREMENT (ABOVE AND BEYOND): Connect toggles to features
      // Here, the .checked property directly manipulates the CSS styles of our page elements!
      
      if (toggleDark.checked) {
        container.style.background = "#1a1a1a";
        container.style.color = "white";
      } else {
        container.style.background = "white";
        container.style.color = "black";
      }

      // Using ternary operators to show/hide the elements based on toggle state
      avatarContainer.style.display = toggleAvatar.checked ? "block" : "none";
      bioContainer.style.display = toggleBio.checked ? "block" : "none";

    });
  });
</script>
```
